#Trying Deep Learning for the first time


###Importing Iris Dataset

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("arshid/iris-flower-dataset")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os

dataset = pd.read_csv(os.path.join(path, "IRIS.csv"))

dataset
```

```python
import numpy as np
import tensorflow as tf
```

##Converting target feature to numbers and separating features



```python
dataset['species'] = dataset['species'].replace({'Iris-setosa': 0,'Iris-versicolor': 1,'Iris-virginica': 2 })
```

```python
X = dataset.drop(columns = ['species'])
Y = dataset['species']
```

##Train/Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state=42)
```

##Feature engineering

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

##Creating, training and evaluating model

```python
from tensorflow import keras

model = tf.keras.Sequential([ tf.keras.layers.Dense(16, activation='relu', input_shape=(4,)),
                             tf.keras.layers.Dense(32, activation='relu'),
                              tf.keras.layers.Dense(3, activation='softmax')])
```

```python
model.compile(tf.keras.optimizers.Adam(),
              loss='sparse_categorical_crossentropy',
              metrics = ['accuracy'])
```

```python
model.fit(X_train, Y_train,batch_size =4, epochs=50)
```

```python
prediction = model.predict(X_test)
print(prediction[0])
i = 0

while i<10:
  if np.argmax(prediction[i]) == 0:
    print('Iris-setosa')
  if np.argmax(prediction[i]) == 1:
    print('Iris-versicolor')
  else:
    print('Iris-virginica')
  i+=1
```
